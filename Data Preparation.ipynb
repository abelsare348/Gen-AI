```python
from pyspark.sql import functions as F
from pyspark.sql.functions import col,concat_ws
```

```python
trips= spark.table("samples.nyctaxi.trips")
```

```python
trips=trips.select(F.concat_ws(" ",col("tpep_pickup_datetime"),col("tpep_dropoff_datetime"),col("trip_distance"),col("fare_amount"),col("pickup_zip"),col("dropoff_zip")).alias("text"))
```

```python
texts = trips.select("text").limit(1000).toPandas()["text"].tolist()
```
